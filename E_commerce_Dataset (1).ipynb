```python
import pandas as pd

orders_df = pd.read_csv("Day9_Orders.csv")
customers_df = pd.read_csv("Day9_Customers.csv")
products_df = pd.read_csv("Day9_Products.csv")

merged_df = orders_df.merge(customers_df, on="Customer_ID", how="left")
merged_df = merged_df.merge(products_df, on="Product_ID", how="left")

split_part1 = merged_df.iloc[: len(merged_df) // 2]
split_part2 = merged_df.iloc[len(merged_df) // 2 :]
processed_df = pd.concat([split_part1, split_part2], ignore_index=True)

processed_df["Order_Date"] = pd.to_datetime(processed_df["Order_Date"])
processed_df["Order_Year"] = processed_df["Order_Date"].dt.year
processed_df["Order_Month"] = processed_df["Order_Date"].dt.month_name()
processed_df["Order_Day"] = processed_df["Order_Date"].dt.day
processed_df["Day_Of_Week"] = processed_df["Order_Date"].dt.day_name()

processed_df["Total_Amount"] = processed_df.apply(
    lambda row: row["Quantity"] * row["Unit_Price"], axis=1
)

processed_df.to_csv("Final_Processed_Ecommerce_Data.csv", index=False)

print("Dataset processed and saved successfully!")
print(f"Total Rows: {processed_df.shape[0]}, Total Columns: {processed_df.shape[1]}\n")
display(processed_df.head())
```
